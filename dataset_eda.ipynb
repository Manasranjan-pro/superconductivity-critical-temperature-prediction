```python

import pandas as pd
import numpy as np
from pathlib import Path

# Your notebook and data folder are in the same project directory
DATA_DIR = Path("data")

train_path = DATA_DIR / "train.csv"
formula_path = DATA_DIR / "unique_m.csv"

# Verify files
print("Training file exists:", train_path.exists())
print("Formula file exists:", formula_path.exists())

# Load datasets
df = pd.read_csv(train_path)
formulas = pd.read_csv(formula_path)

# Display dataset information
print("\nTraining dataset shape:", df.shape)
print("Formula dataset shape:", formulas.shape)

print("\nTraining data:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nFormula data:")
display(formulas.head())
```

```python

# Dataset dimensions
print("Training dataset shape:", df.shape)
print("Formula dataset shape:", formulas.shape)

# Column names and data types
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes.value_counts())

# Preview the data
display(df.head())
display(formulas.head())
```

```python

# Inspect the last five columns
display(df.iloc[:, -5:].head())

# Check whether the expected target exists
target_candidates = [
    col for col in df.columns
    if "critical" in col.lower() or "temp" in col.lower()
]

print("Possible target columns:", target_candidates)
```

```python

print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nMissing values by column:")
display(df.isnull().sum().sort_values(ascending=False).head(15))

print("\nDescriptive statistics:")
display(df.describe().T)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

target = "critical_temp"

if target in df.columns:
    plt.figure(figsize=(9, 5))
    sns.histplot(df[target], bins=40, kde=True)
    plt.xlabel("Critical Temperature (K)")
    plt.ylabel("Number of Materials")
    plt.title("Distribution of Superconducting Critical Temperatures")
    plt.tight_layout()
    plt.show()
else:
    print("Target column not found. Check the column names first.")
```

```python

# Step 2.5: Investigate duplicate records

print("Exact duplicate rows:", df.duplicated().sum())

print("Duplicate feature vectors (excluding target):",
      df.drop(columns=["critical_temp"]).duplicated().sum())

# Inspect material formula column
print("\nFormula dataset columns:")
print(formulas.columns.tolist())

print("\nMaterial formula examples:")
display(formulas[["material", "critical_temp"]].head(10))

# Check whether formulas repeat
print("\nRepeated material formulas:",
      formulas["material"].duplicated().sum())

# Check whether repeated formulas have different targets
target_counts = formulas.groupby("material")["critical_temp"].nunique()
conflicting_formulas = target_counts[target_counts > 1]

print("Formulas with conflicting target values:",
      len(conflicting_formulas))

display(conflicting_formulas.head(10))
```

```python

# Step 2.6: Check target consistency between files

print("Training target equals formula target:",
      np.allclose(
          df["critical_temp"].to_numpy(),
          formulas["critical_temp"].to_numpy(),
          equal_nan=True
      ))

print("\nTarget values in training data:")
display(df["critical_temp"].describe())

print("\nTarget values in formula data:")
display(formulas["critical_temp"].describe())
```

```python

# Step 2.7: Investigate repeated material formulas

# Number of records for each formula
formula_counts = formulas["material"].value_counts()

print("Total records:", len(formulas))
print("Unique formula strings:", formulas["material"].nunique())
print("Formulas occurring more than once:",
      (formula_counts > 1).sum())

print("\nMost frequent formulas:")
display(formula_counts.head(15))

# Examine one formula with multiple target values
example_formula = conflicting_formulas.index[0]

print("\nExample formula:", example_formula)

display(
    formulas.loc[
        formulas["material"] == example_formula,
        ["material", "critical_temp"]
    ].sort_values("critical_temp")
)
```

```python

# Step 2.8: Quantify conflicting labels for identical feature vectors

feature_cols = [
    col for col in df.columns
    if col != "critical_temp"
]

grouped_features = df.groupby(feature_cols)["critical_temp"].agg(
    count="size",
    unique_targets="nunique",
    min_target="min",
    max_target="max",
    mean_target="mean"
)

conflicting_feature_groups = grouped_features[
    grouped_features["unique_targets"] > 1
]

print("Unique feature vectors:", len(grouped_features))
print("Feature vectors with conflicting targets:",
      len(conflicting_feature_groups))

print("\nExamples:")
display(conflicting_feature_groups.head(10))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

target = "critical_temp"
feature_cols = [c for c in df.columns if c != target]

# Feature statistics
feature_stats = df[feature_cols].describe().T
feature_stats["skewness"] = df[feature_cols].skew()

display(
    feature_stats.sort_values(
        "skewness", key=abs, ascending=False
    ).head(15)
)

# IQR-based outlier counts
Q1 = df[feature_cols].quantile(0.25)
Q3 = df[feature_cols].quantile(0.75)
IQR = Q3 - Q1

outlier_counts = (
    (df[feature_cols] < Q1 - 1.5 * IQR) |
    (df[feature_cols] > Q3 + 1.5 * IQR)
).sum()

outlier_summary = outlier_counts.sort_values(ascending=False)

print("Features with the most IQR outliers:")
display(outlier_summary.head(15))

print("Total feature outlier flags:",
      int(outlier_counts.sum()))
```

```python

# Correlation with the target
target_corr = (
    df[feature_cols + [target]]
    .corr(numeric_only=True)[target]
    .drop(target)
)

print("Top 15 positively correlated features:")
display(target_corr.sort_values(ascending=False).head(15))

print("Top 15 negatively correlated features:")
display(target_corr.sort_values().head(15))

# Correlations among the 15 most target-associated features
top_features = target_corr.abs().nlargest(15).index.tolist()

plt.figure(figsize=(12, 9))
sns.heatmap(
    df[top_features + [target]].corr(),
    cmap="coolwarm",
    center=0,
    annot=False
)
plt.title("Correlation Heatmap of Selected Features and Target")
plt.tight_layout()
plt.show()
```

```python

# Select six features with the strongest absolute target correlations
selected_features = target_corr.abs().nlargest(6).index.tolist()

fig, axes = plt.subplots(2, 3, figsize=(16, 9))

for feature, ax in zip(selected_features, axes.flat):
    sns.scatterplot(
        data=df,
        x=feature,
        y=target,
        alpha=0.3,
        s=15,
        ax=ax
    )
    ax.set_title(f"{feature} vs {target}")

plt.tight_layout()
plt.show()
```

```python

# Reuse grouped_features from Step 2.8
conflicts = grouped_features[
    grouped_features["unique_targets"] > 1
].copy()

conflicts["target_range"] = (
    conflicts["max_target"] - conflicts["min_target"]
)

print("Conflicting feature-vector groups:", len(conflicts))
print("Rows belonging to these groups:",
      int(conflicts["count"].sum()))

print("\nTarget-range summary for conflicting groups:")
display(conflicts["target_range"].describe())

print("\nGroups with the largest target differences:")
display(
    conflicts.sort_values("target_range", ascending=False).head(10)
)
```

```python

# Identify elemental columns in the formula dataset
element_cols = [
    c for c in formulas.columns
    if c not in ["critical_temp", "material"]
]

print("Number of elemental columns:", len(element_cols))
print("Number of numerical input features:", len(feature_cols))

# Check whether element values are nonnegative
print(
    "Negative elemental values:",
    int((formulas[element_cols] < 0).sum().sum())
)

# Inspect elemental composition values
display(formulas[element_cols].describe().T)

# Compare the first few row indices and target values
alignment_check = pd.DataFrame({
    "training_target": df["critical_temp"],
    "formula_target": formulas["critical_temp"],
    "material": formulas["material"]
})

display(alignment_check.head(10))
```
